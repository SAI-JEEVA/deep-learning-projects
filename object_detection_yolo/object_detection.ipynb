```python
from ultralytics import YOLO
import torch
```

```python

# 1. Choose the device for inference
device = 0 if torch.cuda.is_available() else "cpu"
print("Using device:", "NVIDIA GPU" if device == 0 else "CPU")

# 2. Load a pretrained YOLO model
model = YOLO("yolo11n.pt")

# 3. Detect objects in a sample image
results = model.predict(
    source="https://ultralytics.com/images/bus.jpg",
    device=device,
    conf=0.25,
    save=True
)

print("Detection complete!")
print("Detected objects:", len(results[0].boxes))
print("Output folder:", results[0].save_dir)
```

### using img from folder ###

```python
from ultralytics import YOLO
import torch
from pathlib import Path

device= 0 if torch.cuda.is_available() else "cpu"

model = YOLO("yolo11n.pt")

img = Path("ob_det-1.webp")

if not img.exists():
    raise FileNotFoundError(
        f"Image not found: {img.resolve()}"
    )

res= model.predict(
    source=str(img),
    device=device,
    conf=0.25,
    save=True
)

print("Number of detections:", len(results[0].boxes))
print("Saved image to:", results[0].save_dir)
```

```python

from ultralytics import YOLO
import torch

# 1. Select GPU if available
device = 0 if torch.cuda.is_available() else "cpu"
print("Using device:", "NVIDIA GPU" if device == 0 else "CPU")

# 2. Load the pretrained YOLO model
model = YOLO("yolo11n.pt")

# 3. Set the input video path
video_path = "videos/solutions_ci_demo.mp4"

# 4. Run object detection on the video
results = model.predict(
    source=video_path,
    device=device,
    conf=0.25,
    save=True,
    stream=True
)

# 5. Process the results frame by frame
frame_count = 0

for result in results:
    frame_count += 1
    detected_objects = len(result.boxes)

    print(
        f"Frame: {frame_count} | "
        f"Detected objects: {detected_objects}"
    )

print("Video processing complete!")
```

```python
from ultralytics import YOLO
import torch

# Step 1: Select the computing device
device = 0 if torch.cuda.is_available() else "cpu"

print("Using device:", "NVIDIA GPU" if device == 0 else "CPU")

# Step 2: Load the pretrained YOLO model
model = YOLO("yolo11n.pt")

# Step 3: Start object detection using the webcam
print("Starting webcam detection...")
print("Press 'q' in the video window to stop.")

results = model.predict(
    source=0,
    device=device,
    conf=0.25,
    show=True,
    stream=True,
    verbose=False
)

# Step 4: Process each frame
frame_count = 0

try:
    for result in results:
        frame_count += 1

        detected_objects = len(result.boxes)

        print(
            f"\rFrame: {frame_count} | "
            f"Objects detected: {detected_objects}",
            end=""
        )

except KeyboardInterrupt:
    print("\nDetection stopped by user.")

finally:
    results.close()
    print("\nWebcam detection ended.")
```
